```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import make_classification
X,y=make_classification(n_samples=1000 ,n_features=2,n_informative=1,n_redundant=0,n_classes=2,n_clusters_per_class=1,random_state=42,
                        hypercube=False,class_sep=10)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,6))
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
```

```python
def perceptron(X,y):              # make a function called perceprtron
  X=np.insert(X,0,1,axis=1)
  weights=np.ones(X.shape[1])
  lr=0.1

  for i in range(1000):                 # size of epochs =1000,
    j = np.random.randint(0,X.shape[0])             ## randomly selected value sbetween the range j represents the value
    y_hat= step(np.dot(X[j],weights))
    weights= weights+lr*(y[j]-y_hat)*X[j]

  return weights[0],weights[1:]
```

```python
def step(z):
  return 1 if z>0 else 0
```

```python
import numpy as np
intercept_,coef_=perceptron(X,y)
```

```python
print(coef_)
print(intercept_)               ## it gives values of w1, w2, wo
```

```python
m= -(coef_[0]/coef_[1])
b = -(intercept_/coef_[1])
```

```python
x_input=np.linspace(-3,3,100)
y_input=m*x_input+ b
y_input
```

```python
plt.figure(figsize=(10,6))
plt.plot(x_input,y_input,color='red',linewidth=3)
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
plt.ylim(-3,2)
```
