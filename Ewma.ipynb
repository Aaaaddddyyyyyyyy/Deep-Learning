```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("/content/DailyDelhiClimate.csv")
```

```python
df.head()
```

```python
df=df[['date','meantemp']]
df
```

```python
plt.scatter(df['date'],df['meantemp'],color='green')
```

```python
x1=df['meantemp'].ewm(alpha=0.9).mean()       ## calcul;ating exponentially moving avereage
x1
```

```python
## merging into dataframe

df['ewm']=x1
df
```

```python
plt.scatter(df['date'],df['meantemp'],color='pink')
plt.plot(df['date'],x1,color='green',label='alpha=0.9')
plt.legend()
plt.show()
```
