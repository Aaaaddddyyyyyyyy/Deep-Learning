```python
def initialize_parameters(layer_dim):

  np.random.seed(3)
  parameters={}
  L=len(layer_dim) # Corrected: layer_dims to layer_dim

  for l in range(1,L):
    parameters['W' + str(l)] = np.ones((layer_dim[l], layer_dim[l-1])) * 0.1 # Corrected syntax, indexing, and np.ones shape
    parameters['b' + str(l)] = np.zeros((layer_dim[l], 1)) # Corrected syntax and indexing


  return parameters
```
