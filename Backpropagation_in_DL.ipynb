```python
def intialize_parameters(layer_dims):
  np.random.seed(3)
  parameters = {}
  L=len(layer_dims) # Corrected: layer_dim -> layer_dims

  for l in range(1,L):
    # Corrected: use 'l' for iteration, and correct dimensions for W and b
    parameters['W'+str(l)]= np.ones((layer_dims[l],layer_dims[l-1]))*0.1
    parameters['b'+str(l)]=np.zeros((layer_dims[l],1))

  return parameters
```

```python
intialize_parameters([2,2,1])    # initialize architecture of neural network
```

```python
def linear_forward(A_prev,W,b):
  Z= np.dot(W.T,A_prev)+b
  return Z
```

```python
## Forward propagation
def L_layer_forward(X,parameters):
  A=X
  L=len(parameters)  //2        ## no of layers in neural network

  for i in range(1,L+1):
    A_prev= A
    W1= parameters['W'+str(1)]
    b1= parameters['b'+str(1)]
    print("A"+str(i-1)+":",A_prev)
    print("W"+str(i)+":",W1)
    print("b"+str(i)+":",b1)
    print("-"*20)


    A=linear_forward(A_prev,W1,b1)
    print("A"+str(i)+": ",A)
  return A,A_prev
```

```python
X=df[['cgpa','profile_score']].values[0].reshape(2,1)
y=df[['lpa']].values[0][0]

# parameter intialize
parameters=intialize_parameters([2,2,1])

L_layer_forward(X,parameters)
```

```python
## update weights and bias
def update_parameters(parameters,y,y_hat,A1,X):
  parameters['W2'][0][0]=parameters['W2'][0][0]+(0.001*2*(y-y_hat)*A1[0][0])
  parameters['W2'][1][0]=parameters['W2'][1][0]+(0.001*2*(y-y_hat)*A1[1][0])
  parameters['b2'][0][0]=parameters['W2'][1][0]+(0.001*2(y-y_hat))

  parameters['W1'][0][0]=parameters['W1'][0][0]+(0.001*2*(y-y_hat)*parameters['W2'][0][0]*X[0][0])
  parameters['W1'][0][1]=parameters['W1'][0][1]+(0.001*2*(y-y_hat)*parameters['W2'][0][0]*X[1][0])
  parameters['b1'][0][0]=parameters['W1'][1][0]+(0.001*2(y-y_hat)*parameters['W2'][0][0]*X[0][0])

  parameters['W1'][1][0]=parameters['W1'][1][0]+(0.001*2*(y-y_hat)*parameters['W2'][1][0]*X[0][0])
  parameters['W1'][1][1]=parameters['W1'][1][1]+(0.001*2*(y-y_hat)*parameters['W2'][1][0]*X[1][0])
  parameters['b1'][1][0]=parameters['b1'][1][0]+(0.001*2(y-y_hat)*parameters['W2'][1][0])
```

```python
## using keras

import tensorflow
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense
```

```python
# Neural network architecture
model= Sequential()

model.add(Dense(2,activation='linear',input_dim=2))
model.add(Dense(1,activation='linear'))
```

```python
## new weights and bias
new_weights=[np.array([[0.1,0.1],[0.1,0.1]],dtype=np.float32),           ## weights=0.1 bias= 0
             np.array([0.,0.],dtype=np.float32),
             np.array([[0.1],[0.1]],dtype=np.float32),
             np.array([0.],dtype=np.float32)]
```

```python
model.set_weights(new_weights)
```

```python
model.get_weights()
```

```python
## setting learning rate
optimizer=keras.optimizers.Adam(learning_rate=0.001)
model.compile(loss='mean_squared_error',optimizer=optimizer)
```

```python
model.fit(df.iloc[:,0:-1].values,df['lpa'].values,epochs=75,verbose=1,batch_size=1)
```
